# Correlations <hr style="border:2.5px solid #e625ffff"></hr>

```python
#imports and packages
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats 
# we can use //import scipy as sp// though that will make memory usage bigger
```

```python
# Load dataset
df = pd.read_csv(r'dataset\current-power.csv', # r read code as is
                 delimiter=",") #delimiter seperates data
df
```

```python
# summary of statistics
df.describe()
```

```python
#covariance of current and power
df.cov()
```

```python
#scatter plot
x = df['Current'] #from the dataframe stores values from current
y = df['Power'] #from the dataframe stores values from power

plt.figure()
plt.scatter(x,y)
plt.title('Device Current vs Power Response')
plt.ylabel('Power (W)')
plt.xlabel('Current (A)')
plt.show() 
```

```python
#correlation coefficient of current and power
df.corr()
```

```python
#regression line
slope, intercept, r_value, p_value, std_err = stats.linregress(x,y)
line = slope*x + intercept

#scatter plot
plt.figure(figsize=(7,4))
plt.scatter(x,y,
            color = "#126782",
            alpha=0.6,
            label='Data points')

#line plot
plt.plot (x,line,
          color = "#126782" ,
          label= f'Regression line (r = {r_value:.3f})') 
#f is the (f string) method
#3f mean decimal places

plt.title('Device Current vs Power Response')
plt.ylabel('Power (W)')
plt.xlabel('Current (A)')
plt.legend()
plt.show()
```

# Air Quality

```python
# Load dataset
air = pd.read_csv(r'dataset\air-quality.csv',
                  delimiter=",")
air
```

# NOx-Humidity

```python
#regression line
x = air['NOx']
y = air['Relative_Humidity']
slope, intercept, r_value, p_value, std_err = stats.linregress(x,y)
line = slope*x + intercept

#scatter plot
plt.figure(figsize=(7,4))
plt.scatter(x,y,
            color="#126782",
            alpha=0.2,
            label='Data points')

#line plot
plt.plot (x,line,
          color="#611CD1" ,
          label= f'Regression line (r = {r_value:.3f})') 
#f is the (f string) method
#3f mean decimal places

plt.title('NOx vs Humidity Correlation')
plt.ylabel('Relative Humidity (%)')
plt.xlabel('NOx (ug/cu.m)')
plt.legend()
plt.show()
```
