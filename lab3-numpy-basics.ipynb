# 2D array

```python
my_array = np.array ([
    [1,2,3]
    [4,5,6]
])
```

```python
np.shape(my_array)
```

```python
my_array.T
#transposes the array
```

```python
np.shape(my_array)
```

```python
trans_array = my_array.T
```

```python
trans_array
```

```python
np.shape(trans_array)
```

```python
np.sum(trans_array, axis=0)
#adds Column
```

```python
np.sum(trans_array, axis=1)
#adds rows
```

```python
resistance = np.array([
    [10.6,9.1,9.3,9.8,10.5,10.4, 9.5, 11,10.4,3,9.8] #jaguar
    [10.1, 11,9.1,20, 9.2,10.8,9.9,9.2,9.1,9.1, np.nan]#panther
])
resistance
```
