```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

# Z-Score


```python
#Creating DataFrame
df = pd.DataFrame([1,2,2,3,3,3,4,4,5],
                columns=['Dataset'])
df
```

```python
# Summary of Statistics
df.describe()
```

```python
# Z-Scores
df['Z-score'] = stats.zscore(df['Dataset'],ddof=1)
df
```

```python
# Load dataset
df = pd.read_csv(r"dataset\current-test.csv",
                 delimiter =',',
                 index_col='Sample')
df

df['Current'][3]

df['Current'][3:5]

sample_1 = df['Current'][3:5].mean()
sample_1

df.info()
```

# Histogram

```python
#histogram plot
plt.hist(df['Current'], bins=5, 
            alpha = 0.8,
            color = "#044E0A",
            edgecolor = '#000000ff')
plt.title('Histogram of Current Values',
          font = 'Palatino Linotype',
          fontsize = 12)
plt.xlabel('Current (mA)')
plt.ylabel('Frequency')
plt.show()
```

# Normal Distribution

```python
# normal distribution curve
mean = df['Current'].mean()
std = df['Current'].std(ddof=1)
#x-axis
x_min = df['Current'].min()
x_max = df['Current'].max()
x = np.linspace(x_min, x_max, 100)
p = stats.norm.pdf(x, mean, std)
plt.plot(x, p,
         color = "#004BA6",
         linewidth = 2)
plt.title('Normal Distribution Current Values',
          font = 'Palatino Linotype',
          fontsize = 12)
plt.xlabel('Current (mA)')
plt.ylabel('Probability Density')
plt.show()
```

```python
#histogram plot
plt.hist(df['Current'], bins=5, 
            density=True,
            alpha = 0.8,
            color = "#0FB100",
            edgecolor = '#000000ff')
plt.title('Histogram of Current Values',
          font = 'Palatino Linotype',
          fontsize = 12)
plt.xlabel('Current (mA)')
plt.ylabel('Frequency')
# normal distribution curve
p = stats.norm.pdf(x, mean, std)
plt.plot(x, p,
         color = "#000000",
         linewidth = 2)
plt.show()
```

```python
# Standard Normal Distribution N(0,1)
# x-axis
x_max = df['Z-score'].max()
x_min = df['Z-score'].min()
x = np.linspace(x_min, x_max, 100) 

p = stats.norm.pdf(x, df['Z-score'].mean(), df['Z-score'].std(ddof=1))

plt.plot(x, p,
         color = "#0FB100",
         linewidth = 2)

plt.title('Standard Normal Distribution of Current Values N(0,1)',
          font = 'Palatino Linotype',
        fontsize = 12)
plt.xlabel('Z-Score')
plt.ylabel('Probability Density')
plt.show()
```
